## Creating Pools
### from_seq

```python
pool = pp.from_seq("AAAA")
library = pool.generate_library()
print(library)
```

```python
wt = pp.from_seq("GCGC<promoter>TATAAA</promoter><core>ATCGATCG</core>GCGC")
print(wt.generate_library())
```

```python
wt      = pp.from_seq("AAAA<cre>ATCG</cre>TTTT")
mutant  = pp.from_seq("GGGG", pool=wt, region="cre")
mutant.generate_library()
#mutant.print_library()
```

### from_seqs

```python
pool = pp.from_seqs(["AAAA", "CCCC", "GGGG", "TTTT"], num_states=4)
pool.generate_library()
```

```python
pool = pp.from_seqs(
    ["ATCG", "CGTA", "GCAT"],
    mode="sequential",
)
pool.generate_library()
```

```python
pool = pp.from_seqs(
    ["ATCG", "CGTA", "GCAT"],
    seq_names=["wt", "mut_A", "mut_B"],
    mode="sequential",
)
pool.generate_library()
```

### from_fasta

```python
pool = pp.from_fasta(
    "genome.fa",
    coordinates=("chr1", 1000, 1020, "+"),
)
pool.generate_library()
```

```python
pool = pp.from_fasta(
    "genome.fa",
    coordinates=("chr1", 5000, 5010, "-"),
)
pool.generate_library()
```

```python
coords = [
    ("chr1", 1000, 1010, "+"),
    ("chr1", 5000, 5010, "-"),
    ("chr1", 200,  210,  "+"),
]
pool = pp.from_fasta("genome.fa", coordinates=coords, prefix="enh")
pool.generate_library()
```

```python
vector = pp.from_seq("GCGCGC<insert>XXXXXXXXXX</insert>GCGCGC")
coords = [("chr1", 1000, 1010, "+"), ("chr1", 2000, 2010, "+")]
pool   = pp.from_fasta("genome.fa", coordinates=coords,
                       pool=vector, region="insert")
pool.generate_library()
```

### from_motif

```python
import pandas as pd

pfm = pd.DataFrame(
    {"A": [0.25, 0.25], "C": [0.25, 0.25],
     "G": [0.25, 0.25], "T": [0.25, 0.25]}
)
pool = pp.from_motif(pfm, num_states=4)
```

```python
pool.generate_library(num_cycles=2, seed=42)
```

```python
pfm = pd.DataFrame(
    {"A": [0.7, 0.1], "C": [0.1, 0.7],
     "G": [0.1, 0.1], "T": [0.1, 0.1]}
)
bg   = pp.from_seq("GCGC<insert>XX</insert>GCGC")
pool = pp.from_motif(pfm, pool=bg, region="insert", num_states=4)
pool.generate_library()
```

## Combining Pools

### join

```python
left  = pp.from_seqs(["AA", "CC"], mode="sequential")
right = pp.from_seqs(["GG", "TT"], mode="sequential")   
pool  = pp.join([left, right])
pool.generate_library()
```

```python
left  = pp.from_seq("ACGT")
right = pp.from_seqs(["AAAA", "CCCC", "GGGG"], mode="sequential")
pool  = pp.join([left, right], spacer_str="TTT")
pool.generate_library()
```

```python
a    = pp.from_seqs(["A", "C"], mode= 'sequential', iter_order=2)   # outer (slower)
b    = pp.from_seqs(["G", "T"], mode='sequential', iter_order=1)   # inner (faster)
pool = pp.join([a, b])
pool.generate_library()
```

```python
promoter = pp.from_seqs(["AAAA", "CCCC"], mode="sequential")
insert   = pp.from_seqs(["GCGC", "TATA"], mode="sequential")
pool     = pp.join([promoter, "NNNNN", insert])
pool.generate_library()
```

### Concatenation operators

```python
wt   = pp.from_seq("ATCG")
muts = pp.mutagenize(wt, num_mutations=1, mode="sequential")
ctrl = pp.from_seqs(["AAAA", "TTTT"], mode="sequential")
lib  = muts + ctrl   # all single-point mutants + 2 controls
lib.generate_library()
muts.generate_library()
```

```python
unit = pp.from_seqs(["AT", "CG"], mode="sequential")
pool = unit * 3
pool.generate_library()
```
