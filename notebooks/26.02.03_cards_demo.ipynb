```python
import poolparty as pp; pp.init()

print("=== Default output: only name and seq columns ===")
pool = pp.from_seqs(["ACGT", "TGCA", "GGGG"], mode="sequential")
pool.mutagenize(num_mutations=1).generate_library(num_cycles=1).pipe(print)

print("\n=== Requesting design card with list (default column naming) ===")
pool = pp.from_seqs(["ACGT", "TGCA"], mode="sequential", cards=["seq_index"])
pool.generate_library(num_cycles=1).pipe(print)

print("\n=== Requesting design card with dict (custom column naming) ===")
pool = pp.from_seqs(["ACGT", "TGCA"], mode="sequential", cards={"seq_index": "my_idx"})
pool.generate_library(num_cycles=1).pipe(print)

print("\n=== Universal cards: seq and state (available on all operations) ===")
pool = pp.from_seqs(["ACGT", "TGCA"], mode="sequential", cards=["seq", "state"])
pool.generate_library(num_cycles=1).pipe(print)

print("\n=== Multiple operations with different cards ===")
pool = pp.from_seqs(["ACGT", "TGCA"], mode="sequential", cards=["seq_index"])
mutated = pool.mutagenize(num_mutations=1, cards=["positions", "wt_chars", "mut_chars"])
mutated.generate_library(num_cycles=1).pipe(print)

print("\n=== Custom naming on mutagenize ===")
pool = pp.from_seqs(["ACGT"], mode="sequential")
mutated = pool.mutagenize(num_mutations=1, cards={"positions": "mut_pos", "mut_chars": "new_bases"})
mutated.generate_library(num_seqs=3, seed=42).pipe(print)

print("\n=== Using factory functions directly (from_iupac with cards) ===")
iupac_pool = pp.from_iupac("NNNN", mode="sequential", num_states=4, cards=["iupac_state"])
iupac_pool.generate_library(num_cycles=1).pipe(print)
```
